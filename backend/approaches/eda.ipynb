```python
import pandas as pd
from googletrans import Translator

# Load the CSV file
# file_path = "your_file.csv"  # Replace with your file path
# df = pd.read_csv(file_path)

# Initialize the translator
translator = Translator()

# Define columns to translate
columns_to_translate = ['project_name', 'district', 'unit_type', 'apartment_type']

# Translate the columns
for column in columns_to_translate:
    if column in file.columns:
        col = column + "new"
        file[col] = file[column].apply(lambda x: translator.translate(str(x), src='ar', dest='en').text if pd.notnull(x) else x)

# Save the updated file
output_file = r"C:\Users\rahul\Desktop\Offshore\PubSec-Info-Assistant-Offshore\data\translated_file.csv"
file.to_csv(output_file, index=False)

print(f"Translation completed. Saved to {output_file}")
```
